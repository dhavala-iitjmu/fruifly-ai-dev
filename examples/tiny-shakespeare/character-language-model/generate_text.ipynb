# Generate Tiny Shakespeare
Load the trained MLX connectome language model and sample text at several temperatures.

```python
from pathlib import Path
import json
from generate import load, generate

RUN = Path('runs/malecns-frozen-adapter')
model, meta, config = load(RUN)
print(f"Backend: {config['backend']}")
print(f"Graph: {config['graph']}")
print(f"Nodes: {config['neurons']:,}; vocabulary: {config['vocab_size']}")
print(f"Best validation perplexity: {__import__('math').exp(config['best_val_loss']):.2f}")
```

```python
PROMPT = 'ROMEO:\n'
print(generate(model, meta, PROMPT, length=500, temperature=0.8, top_k=20, seed=7))
```

```python
for temperature in (0.5, 0.8, 1.1):
    print(f'\n--- temperature={temperature} ---')
    print(generate(model, meta, PROMPT, length=250, temperature=temperature, top_k=20, seed=7))
```
